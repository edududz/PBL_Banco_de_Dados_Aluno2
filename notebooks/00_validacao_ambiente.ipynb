```python
import pandas as pd
from database import conectar

sql = """
SELECT
    schemaname AS esquema,
    tablename AS tabela
FROM pg_tables
WHERE schemaname = 'industria'
ORDER BY tablename;
"""

with conectar() as conn:
    with conn.cursor() as cur:
        cur.execute(sql)
        dados = cur.fetchall()

tabelas = pd.DataFrame(
    dados,
    columns=["Esquema", "Tabela"]
)

display(tabelas)

print(f"Total de tabelas: {len(tabelas)}")
```

```python
sql = """
SELECT COUNT(*) AS total_movimentacoes
FROM industria.movimentacoes;
"""

with conectar() as conn:
    with conn.cursor() as cur:
        cur.execute(sql)
        total = cur.fetchone()[0]

print(f"Total de movimentações: {total:,}".replace(",", "."))
```
